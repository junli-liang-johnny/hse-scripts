```python
# Install required packages
%pip install pandas
%pip install requests

import pandas as pd
import re
import io

# Alternative approach to read PX files
def parse_px_file(filepath):
    """
    Simple PX file parser to extract data and metadata
    """
    with open(filepath, 'r', encoding='utf-8') as f:
        content = f.read()
    
    # Split into metadata and data sections
    lines = content.split('\n')
    metadata = {}
    data_started = False
    data_lines = []
    
    for line in lines:
        line = line.strip()
        if not line:
            continue
            
        if line.startswith('DATA='):
            data_started = True
            continue
            
        if data_started:
            if line.endswith(';'):
                data_lines.append(line[:-1])  # Remove semicolon
                break
            else:
                data_lines.append(line)
        else:
            # Parse metadata
            if '=' in line:
                key_part = line.split('=')[0].strip()
                value_part = '='.join(line.split('=')[1:]).strip()
                if value_part.endswith(';'):
                    value_part = value_part[:-1]
                if value_part.startswith('"') and value_part.endswith('"'):
                    value_part = value_part[1:-1]
                metadata[key_part] = value_part
    
    return metadata, data_lines

```

```python
import os

# List all PX files in the temp_datasets folder
px_files = [f for f in os.listdir('temp_datasets') if f.endswith('.px')]

# Parse metadata for each PX file
all_metadata = []
for px_file in px_files:
  meta, _ = parse_px_file(os.path.join('temp_datasets', px_file))
  all_metadata.append((px_file, meta))

# Get the set of all metadata keys across all files
all_keys = set()
for _, meta in all_metadata:
  all_keys.update(meta.keys())

print(f"Total PX files analyzed: {len(px_files)}")
print(f"Total unique metadata keys found: {len(all_keys)}")
print("\nAll metadata keys across files:")
for key in sorted(all_keys):
    print(f"  - {key}")

# Find differences in metadata keys
reference_file, reference_meta = all_metadata[0]
print(f"\nUsing '{reference_file}' as reference file")

# Track files with different keys
files_with_different_keys = []
for px_file, meta in all_metadata[1:]:
    if set(meta.keys()) != set(reference_meta.keys()):
        files_with_different_keys.append(px_file)
        missing_keys = set(reference_meta.keys()) - set(meta.keys())
        extra_keys = set(meta.keys()) - set(reference_meta.keys())
        
        print(f"\n{px_file} has different metadata structure:")
        if missing_keys:
            print(f"  Missing keys: {sorted(missing_keys)}")
        if extra_keys:
            print(f"  Extra keys: {sorted(extra_keys)}")

# Find differences in metadata values for files with same structure
print(f"\n{'='*50}")
print("METADATA VALUE DIFFERENCES")
print(f"{'='*50}")

# Group files by their metadata structure
structure_groups = {}
for px_file, meta in all_metadata:
    key_signature = tuple(sorted(meta.keys()))
    if key_signature not in structure_groups:
        structure_groups[key_signature] = []
    structure_groups[key_signature].append((px_file, meta))

# Analyze each structure group
for i, (key_signature, files_in_group) in enumerate(structure_groups.items()):
    print(f"\nStructure Group {i+1} ({len(files_in_group)} files):")
    print(f"Keys: {list(key_signature)}")
    
    if len(files_in_group) > 1:
        # Compare values within this structure group
        ref_file, ref_meta = files_in_group[0]
        print(f"Reference file: {ref_file}")
        
        for px_file, meta in files_in_group[1:]:
            differences = []
            for key in key_signature:
                if key in ref_meta and key in meta and ref_meta[key] != meta[key]:
                    differences.append((key, ref_meta[key], meta[key]))
            
            if differences:
                print(f"\n  {px_file} differs from reference:")
                for key, ref_val, file_val in differences:
                    print(f"    {key}:")
                    print(f"      Reference: {ref_val}")
                    print(f"      This file: {file_val}")
            else:
                print(f"  {px_file}: No differences")
    else:
        print(f"  Only file: {files_in_group[0][0]}")

# Summary
print(f"\n{'='*50}")
print("SUMMARY")
print(f"{'='*50}")
print(f"Total files: {len(all_metadata)}")
print(f"Different metadata structures: {len(structure_groups)}")
if len(structure_groups) == 1:
    print("✓ All files have the same metadata structure")
else:
    print("✗ Files have different metadata structures")
    for i, (_, files_in_group) in enumerate(structure_groups.items()):
        print(f"  Group {i+1}: {len(files_in_group)} files")
```
